```python
# Pivot years as columns
sales_trend = product_yearly_sales.pivot(
    index='Name',
    columns='Year',
    values='LineTotal'
)

# Check if sales increased every year
increasing_products = sales_trend[
    sales_trend.diff(axis=1).iloc[:,1:].gt(0).all(axis=1)
]

increasing_products.head(5)
```

```python
# Select top 10 growing products based on total revenue
top_growth_products = (
    increasing_products.sum(axis=1)
    .sort_values(ascending=False)
    .head(10)
)

# Create pie chart
plt.figure(figsize=(6,4))

plt.pie(
    top_growth_products.values,
    labels=top_growth_products.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Revenue Contribution of Products with Increasing Sales")

plt.show()
```
